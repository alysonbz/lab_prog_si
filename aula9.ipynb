#DICIONÁRIO DE DICIONÁRIO

1) Acessando os elementos do dicionário europe, print ‘paris’ que pertece ao subdicionário ‘france’.

2) Crie um novo dicionário chamado data e inclua neste 'capital' e 'population'. Atribua, respectivamente 'rome' e 59.83.

3) Crie uma nova chave ‘Italy’ para europe e inclua data dentro da chave.

4) Print europe.


```python
#DEFINITION OF DICTIONARIES
europe = {'spain': {'capital':'madrid', 'population':46.77},
          'france': {'capital':'paris', 'population':66.03},
          'germany': {'capital':'berlin', 'population':80.62},
          'norway': {'capital':'oslo', 'population':5.084}}

#PRINT OUT THE CAPITAL OF FRANCE
print(europe['france']['capital'])

#CREATE SUB-DICTIONARY DATA
data = {'capital':'rome', 'population':59.83}

#ADD DATA TO EUROPE UNDER KEY 'ITALY'
europe['italy'] = data

#PRINT EUROPE
print(europe)
```
