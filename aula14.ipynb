```python
import pandas as pd
brics = pd.read_csv("brics.csv", index_col = 0)
for lab, row in brics.iterrows():
  # - creating series on every iteration
  brics.loc[lab, "name_length"] = len(row["country"])
print(brics)
```

```python
# DESAFIO (questao semelhante na prova?)
# faca uma funcao check.area
# retorta duas possibilidades:
# pais pequeno, se pais menor que 8
#pais grande, caso contrário criar uma nova coluna, country_size
#pais grande ou pais pequeno

import pandas as pd
brics = pd.read_csv("brics.csv", index_col = 0)

def check_area(area):
  if area < 8:
    return"pais pequeno"
  else:
    return"pais grande"
```

```python
brics["tamanho"] = brics["area"].apply(check_area)
```
