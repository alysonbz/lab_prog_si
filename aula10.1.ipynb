*Pandas - Acesso às linhas pela função loc*

```python
brics.loc['RU']
```

```python
brics.loc[['RU']] #irá deixar tudo em uma só linha.
```

*Pandas - cont.: Acesso a linhas*

```python
brics.loc[['RU', 'IN', 'CH']]
```

*Pandas - Acesso a linhas e colunas*

```python
brics.loc[['RU', 'IN', 'CH'], ['country', 'capital']]
```

```python
brics.loc[:, ['country', 'capital']]
```

*Pandas - Acesso à linhas pelo iloc*

```python
brics.iloc[[1]] #daria o mesmo resultado se fizesse -> brics.loc[['RU']]
```

```python
brics.iloc[[1, 2, 3]]
```

*Pandas - Acesso a linhas e colunas pelo iloc*

```python
brics.iloc[[1, 2, 3], [0, 1]]
```
