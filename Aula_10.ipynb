LOOP WHILE

```python
#1)crie uma variável offset. Inicialize ela igual 10.

offset = 10

#Utilize o looop while para processar enquanto a variável offset for diferente de zero.
#Dentro do loop while decrementa a variável offset em 1 e print esta.
while offset != 0:
  offset -= 1
  print(offset)
```

LOOP EM NUMPY

```python
import numpy as np
np_height = np.array ([1.73, 1.68, 1.71, 1.89, 1.79])
np_weight = np.array([65.4, 59.2, 63.6, 88.4, 68.7])
bmi = np_weight/np_height**2
for val in bmi:
  print(val)
```

```python
import numpy as np

np_height = np.array ([1.73, 1.68, 1.71, 1.89, 1.79])
np_weight = np.array([65.4, 59.2, 63.6, 88.4, 68.7])
meas = np.array ([np_height, np_weight])
for val in meas:
  print(val)
```

```python
import numpy as np

np_height = np.array ([1.73, 1.68, 1.71, 1.89, 1.79])
np_weight = np.array([65.4, 59.2, 63.6, 88.4, 68.7])
meas = np.array ([np_height, np_weight])
for val in np.nditer(meas): #np.nditer falar um por um
  print(val)
```
