#CUSTOMIZAÇÃO DE GRÁFICOS

*PLOT BÁSICO*

```python
import matplotlib.pyplot as plt

year = [1600, 1700, 1880, 1990, 2000, 2100]
pop = [2.51, 2.57, 2.83, 2.89, 2.95, 3.0]

plt.plot(year, pop)
plt.show()
```

*INCLUSÃO DE LEGENDAS NOS EIXOS*

```python
import matplotlib.pyplot as plt

year = [1600, 1700, 1880, 1990, 2000, 2100]
pop = [2.51, 2.57, 2.83, 2.89, 2.95, 3.0]

plt.plot(year, pop)

plt.xlabel('Year') #legenda
plt.ylabel('Population') #legenda

plt.show()
```

*INCLUSÃO DE TÍTULOS*

```python
import matplotlib.pyplot as plt

year = [1600, 1700, 1880, 1990, 2000, 2100]
pop = [2.51, 2.57, 2.83, 2.89, 2.95, 3.0]

plt.plot(year, pop)

plt.xlabel('Year')
plt.ylabel('Population')
plt.title('World Population Projections') #título

plt.show()
```

*YTICKS E XTICKS - MARCAÇÕES NOS EIXOS*

```python
import matplotlib.pyplot as plt

year = [1600, 1700, 1880, 1990, 2000, 2100]
pop = [2.51, 2.57, 2.83, 2.89, 2.95, 3.0]

plt.plot(year, pop)

plt.xlabel('Year')
plt.ylabel('Population')
plt.title('World Population Projections')
plt.yticks([0, 2, 4, 6, 8, 10],
            ['0', '2B', '4B', '6B', '8B', '10B'])

plt.show()
```

```python
import matplotlib.pyplot as plt

year = [1600, 1700, 1880, 1990, 2000, 2100]
pop = [2.51, 2.57, 2.83, 2.89, 2.95, 3.0]

#ADD MORE DATA
year = [1800, 1850, 1900] + year
pop = [1.0, 1.262, 1.650] + pop

plt.plot(year, pop)

plt.xlabel('Year')
plt.ylabel('Population')
plt.title('World Population Projections')
plt.yticks([0, 2, 4, 6, 8, 10],
            ['0', '2B', '4B', '6B', '8B', '10B'])

plt.show()
```
