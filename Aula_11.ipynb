```python
#Com auxilio do loop for mostre cada coluna separadamente do dataframe cars.

import pandas as pd
cars = pd.read_csv('/content/sample_data/cars (1).csv', index_col = 0 )

#Com auxílio do loop for realize uma iteraçao nas linhas mostrando a linha id e os elemento contidos de cada linha do dataframe.
for coluna in cars.columns:
    print(f"Coluna: {coluna}")
    print(cars[coluna])
    print()


#Com auxílio do loop for realize uma iteraçao nas linhas mostrando a linha id e os elemento contidos de cada linha do dataframe.
for index, row in cars.iterrows():
    print(f"ID: {index}")
    print(row)
```

```python
#com auxílio do loop for crie uma nova coluna COUNTRY que possui os nomes do países com letra maiúscula.
#dica: Lembre da função upper() para strings.

import pandas as pd
cars = pd.read_csv('/content/sample_data/cars (1).csv', index_col = 0 )

country_names_upper = []

for label in cars.index:
  country_names_upper.append(label.upper())

cars['COUNTRY'] = country_names_upper
print(cars)
```
