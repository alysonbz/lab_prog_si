```python
import numpy as np

altura = np.array([1.73, 1.68, 1.71, 1.89, 1.79])
peso = np.array([65.4, 59.2, 63.6, 88.4, 68.7])
imc = peso / altura ** 2
```

```python
if imc.all() > 30:
    print('Sobrepeso')
    print(imc[imc > 30])
else:
    print('Não está acima do peso')
    print(imc[imc < 30])
```
