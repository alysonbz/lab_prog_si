```python
#questão 8
import pandas as pd
df = pd.DataFrame({"alunos": ["ana", "beto", "carlos", "diana"],
                   "notas": [8, 6, 9, 7]})

for i in range(len(df)):
    print(f"{df['alunos'][i]} tirou nota {df['notas'][i]}")
```

```python
#questão 15
def pegar_pares(lista):
    pares = []
    for num in lista:
        if num % 2 == 0:
            pares.append(num)
    return pares

print(pegar_pares([1, 2, 3, 4, 5, 6]))
```
