```python
#Para 1 valor de peso e um valor de altura, print na tela: Você está com sobrepeso se bmi > 30, caso contrário, você esta com peso adequado.
import numpy as np
np_height = np.array([1.73, 1.68, 1.71, 1.89, 1.79])
np_weight = np.array([65.4, 59.2, 63.6, 88.4, 68.7])
bmi = np_weight / np_height ** 2
for val in bmi:
  if val > 30:
    print('você está acima do peso')
  else:
    print('peso adequado')
```

```python
#Para o Array resposta do exemplo ao lado print, usando operador lógico, onde bmi é menor que 23.
print(bmi[bmi>23])
```

```python
import numpy as np
my_house = np.array([18.0, 20.0, 10.75, 9.50])
your_house = np.array([14.0, 24.0, 14.25, 9.0])
```

```python
#verifique quais elementos de my_house são maiores que 18.5 ou menores que 10.
my_house[np.logical_or(my_house > 18.5, my_house < 10)]
```

```python
#Verifique quis elementos das mesmas posições dos arrays possuem valor menor que 11.
for item in np.logical_and(my_house < 11, your_house < 11):
  print(item)
```
