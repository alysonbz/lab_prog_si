```python
import pandas as pd
```

```python
brics = pd.read_csv('brics.csv', index_col=0)
```

```python
for val in brics:
    print(val)
```

```python
for lab, row in brics.iterrows():
    print(lab)
    print(row)
```

```python
for lab, row in brics.iterrows():
    print(lab + ': ' + row['capital'])
```

```python
for lab, row in brics.iterrows():
  brics.loc[lab, 'name_length'] = len(row['country'])
print(brics)
```

```python
brics['name_length'] = brics['country'].apply(len)
print(brics)
```

```python
#atividade check_area
```

```python
def check_area(area):
    is_big = "pequena"
    if area > 8:
        is_big = "grande"
    return is_big
```

```python
brics['is_huge'] = brics['area'].apply(check_area)
print(brics)
```

```python
for lab, row in brics.iterrows():
  print(f"{lab}: {row['population']}")
```

```python
for lab, row in brics.iterrows():
  brics.loc[lab, 'country'] = row['country'].upper()
print(brics)
```
